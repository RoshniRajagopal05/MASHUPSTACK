```python
import pandas as pd
data = {
    'Team': ['A', 'A', 'B', 'B', 'A'],
    'Gender': ['M', 'F', 'M', 'F', 'M'],
    'Score': [88, 92, 85, 90, 75]
}
df = pd.DataFrame(data)

# Group by one column
print(df.groupby('Team')['Score'].mean())

# Group by multiple columns
df.groupby(['Team', 'Gender'])['Score'].mean()
```

```python
import pandas as pd
data = {
    'Team': ['A', 'A', 'B', 'B', 'A'],
    'Gender': ['M', 'F', 'M', 'F', 'M'],
    'Score': [88, 92, 85, 90, 75]
}
df = pd.DataFrame(data)

def range_func(series):
    return series.max() - series.min()

grouped = df.groupby('Team')['Score'].agg(range_func)
print(grouped)
```

```python
import pandas as pd
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [50000, 40000, 55000, 42000]
})
result = df.groupby("Department").agg({
    "Salary": ["sum", "mean"]
})
print(result)
```

```python
import pandas as pd
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [50000, 40000, 55000, 42000]
})
result = df.groupby("Department").agg(
    Total_Salary=("Salary", "sum"),
    Average_Salary=("Salary", "mean")
)
print(result)
```

```python
import pandas as pd
data = {
    'Department': ['HR', 'HR', 'IT', 'IT', 'Finance'],
    'Employee': ['A', 'B', 'C', 'D', 'E'],
    'Salary': [3000, 3200, 4000, 4100, 5000]
}
df = pd.DataFrame(data)
pivot = df.pivot_table(values='Salary', index='Department', aggfunc='mean')
print(pivot)
```
