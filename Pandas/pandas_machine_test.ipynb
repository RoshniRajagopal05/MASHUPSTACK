```python
print(df["Purchase_Date"])
df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"])
print(f" Datatype_of_purchase_date : {df["Purchase_Date"].dtype}")
```

```python
renaming_payment_mode = df.rename(columns={"Payment_Mode" : "PaymentType"})
print(renaming_payment_mode)
```

```python
df["Total_amount"] = df["Price"] * df["Quantity"]
print(df[["Price", "Quantity", "Total_amount"]].head())
```

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins = [0, 25, 35, 50, float("inf")],
    labels=["18-25", "26-35", "36-50", "51+"]

)
```

```python
print(df[["Age", "Age_Group"]].head(20))
```

```python
print(df.dtypes)
```

```python
df["Customer_ID"] = df["Customer_ID"].astype("int32")
df["Age"] = df["Age"].astype("int8")
df["Quantity"] = df["Quantity"].astype("int8")
df["Price"] = df["Price"].astype("int32")
```

```python
df["Total_amount"] = df["Total_amount"].astype("int32")
```

```python
df["Gender"] = df["Gender"].astype("category")
df["City"] = df["City"].astype("category")
df["Product"] = df["Product"].astype("category")
if "PaymentType" in df.columns:
    df["PaymentType"] = df["PaymentType"].astype("category")
elif "Payment_Mode" in df.columns:
    df["Payment_Mode"] = df["Payment_Mode"].astype("category")
else:
    raise KeyError("Neither 'PaymentType' nor 'Payment_Mode' exists in df.")
df["Age_Group"] = df["Age_Group"].astype("category")
```

```python
city_analysis = (
    df.groupby("City").agg(
        Total_Sales = ("Total_amount", "sum"),
        Average_Price = ("Price", "mean"),
        Total_Quantity = ("Quantity", "sum")
    )
)

print(city_analysis)
```

```python
df["Year"] = df["Purchase_Date"].dt.year
df["Month"] = df["Purchase_Date"].dt.month

print(df[["Purchase_Date", "Year", "Month"]])
```

```python
sales_pivot = pd.pivot_table(
    df,
    values="Total_amount",
    index= "City",
    columns= "Product",
    aggfunc="sum"
)
print("Pivot Table for Sales")
print("Sales by city and Product : ")
print(sales_pivot)
```

```python
quantity_pivot = pd.pivot_table(
    df,
    values="Quantity",
    index="City",
    columns="Product",
    aggfunc="sum"
)

print("\n Pivot table for Quantity")
print("\n Quantity sold by City and Product :")
print(quantity_pivot)
```

```python
payment_crosstabs = pd.crosstab(
    df["City"],
    df["Payment_Mode"]
)
print("City vs Payment Type :")
print(payment_crosstabs)
```

```python
product_gender_crosstab = pd.crosstab(
    df["Product"],
    df["Gender"]
)

print("\n Product vs Gender")
print(product_gender_crosstab)
```

```python
monthly_sales = (
    df.groupby(["Year", "Month"], as_index=False)["Total_amount"].sum()
)
print( monthly_sales)
```

```python
monthly_sales_pivot = pd.pivot_table(
    df,
    values="Total_amount",
    index="Month",
    columns="Year",
    aggfunc="sum"
)
print("\n Monthly Sales Pivot Table :")
print(monthly_sales_pivot)
```
