```python
import pandas as pd
data = {
    'Department': ['HR', 'HR', 'IT', 'IT', 'Finance'],
    'Employee': ['A', 'B', 'C', 'D', 'E'],
    'Salary': [3000, 3200, 4000, 4100, 5000]
}
df = pd.DataFrame(data)
pivot = df.pivot_table(values='Salary', index='Department', aggfunc='mean')
print(pivot)
```

```python
import pandas as pd
data = {
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR']
}
df = pd.DataFrame(data)
ct = pd.crosstab(df['Gender'], df['Department'])
print(ct)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Asha", "Ravi"],
    "Math": [85, 90],
    "Science": [78, 88]
})

result = pd.melt(df, id_vars="Name", var_name="Subject", value_name="Marks")
print(result)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Math": [85, 90],
    "Science": [78, 88]
}, index=["Asha", "Ravi"])

result = df.stack()
print(result)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Math": [85, 90],
    "Science": [78, 88]
}, index=["Asha", "Ravi"])

stacked = df.stack()
result = stacked.unstack()
print(result)
```
