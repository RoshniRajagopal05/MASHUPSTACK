```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Anu"],
    "Age": [23, 25, 24],
    "Salary": [35000, 30000, 40000]
})

print(df)
```

```python
df.to_json("employees.json")
```

```python
df.to_json("Data/employees.json")
```

```python
import os
import pandas as pd

data = {'City': ['New York', 'Paris'], 'Temperature': [22, 18]}
df = pd.DataFrame(data)

df.to_json('climate.json')

file_path = os.path.abspath('climate.json')
print(f"File saved at: {file_path}")
```
