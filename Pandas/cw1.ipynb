```python
import pandas as pd
sales_series = pd.Series([250, 300, 150, 400, 350],
index=["Mon", "Tue", "Wed","Thurs", "Fri"]

                 )
print(sales_series)
print(sales_series["Wed"])
print(sales_series.sum())
print(sales_series.mean())
```

```python
sales_df = pd.DataFrame({
    "Day" : ["Mon", "Tue", "Wed", "Thurs", "Fri"],
    "Sales" : [250, 300, 150, 400, 350 ]
})
print(sales_df)
print("Minimum Sales : ", sales_df["Sales"].min())
print("Maximum Sales : ", sales_df["Sales"].max() )
```
