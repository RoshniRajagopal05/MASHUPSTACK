```python
import pandas as pd

student1_marks = pd.Series([85, 78, 90, 88],
index = ["Math", "Science", "English", "Computer"])

print(student1_marks)
print("Highest Marks : ", student1_marks.max())
print("Lowest Marks : ", student1_marks.min())
print("Total_marks : ", student1_marks.sum())
print("Average marks : ", student1_marks.mean())

```

```python
marks_df = pd.DataFrame({
    "Name" : ["Arun", "Binu", "Chitra"],
    "Math" : [85, 70, 92 ],
    "Science" : [78, 82, 88],
    "English" : [90, 76, 95],
    "Computer" : [ 88, 80, 91]

})

print(marks_df)
marks_df["Total"] = marks_df[["Math", "Science", "English", "Computer"]].sum(axis=1)
print(marks_df)
```

```python
marks_df["Average"] = marks_df[["Math", "Science", "English", "Computer"]].mean(axis=1)
print(marks_df)
```

```python
print("Highest Total marks :", marks_df[["Name","Total"]].max())
```
