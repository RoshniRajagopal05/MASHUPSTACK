```python
import sqlite3

com = sqlite3.connect("school.db")
com.execute("""
CREATE TABLE IF NOT EXISTS students(
    ID INTEGER,
    Name TEXT,
    Mark INTEGER
)
""")
```

```python
com.execute("INSERT INTO students VALUES (1, 'Arun', 85)")
com.execute("INSERT INTO students VALUES (2, 'Binu', 78)")
com.execute("INSERT INTO students VALUES (3, 'Chitra', 92)")

com.commit()
```

```python
students_df = pd.read_sql("SELECT *  FROM students", com)
print(students_df)
```
