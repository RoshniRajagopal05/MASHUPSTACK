```python
import numpy as np
print(np.__version__)
```

```python
marks = np.array([50, 60, 70 , 80 , 90])
print(marks);
print(marks.shape)
```

```python
marks_2d = np.array([
    [30,40,50,60],
    [75,85,95,55]
])

print(marks_2d)
print(marks_2d.shape)
print(marks_2d.ndim)
print(marks_2d.dtype)
print(marks_2d.size)
```

```python
marks_3d = np.array([
[
    [30,50,70],
    [85, 45, 55]
],
[
    [50, 70, 80],
    [22, 56, 88]
]
])
print(marks_3d)
print(marks_3d.shape)
```

```python
zeros_array = np.zeros((3,4))
print(zeros_array)
```

```python
ones_array = np.ones((2,3))
print("\nOnes:")
print(ones_array)
```

```python
five_array = np.full((3,3),5)
print("\nFilled with 5:")
print(five_array)
```

```python
arr = np.arange(10,51,10)
print("Using arange:")
print(arr)
```

```python
arr = np.linspace(0, 100, 5)
print("\nUsing linspace:")
print(arr)
```

```python
identity_matrix = np.eye(4)
print("Identity Matrix:")
print(identity_matrix)
```

```python
diagonal_matrix = np.diag([10, 20, 30, 40])
print("\nDiagonal Matrix:")
print(diagonal_matrix)
```
