```python
import numpy as np

one_d_array = np.array([75, 82, 90, 68, 95])
print("One_d_array", one_d_array)
```

```python
two_d_array = np.array([
    [75, 80, 85],
    [90, 88, 95]
])
print("Two-d array : ")
print(two_d_array)
```

```python
three_d_array = np.array([
    [  [75, 88, 95],
       [65, 55, 45]
    ],

    [
        [99, 33, 78],
        [90, 85, 25]
    ]
])

print("\nThree-d array :")
print(three_d_array)
```

```python
print("Shape :", two_d_array.shape)
print("Dimensions : ", two_d_array.ndim)
print("Datatype : ", two_d_array.dtype)
print("Size : ", two_d_array.size)
```

```python
zeros_array = np.zeros((3,5))
print(zeros_array)
```

```python
ones_array = np.ones((3,5))
print(ones_array)
```

```python
filled_seven = np.full((3,3),5)
print(filled_seven)
```

```python
arr = np.arange(5, 50, 10)
print("Using arange :")
print(arr)
```

```python
arr = np.linspace(0, 100, 6)
print("Using linspace")
print(arr)
```

```python
identity_matrix = np.eye(5)
print(identity_matrix)
```

```python
diagonal_matrix_ = np.diag([5, 10, 15, 20])
print(diagonal_matrix_)
```
